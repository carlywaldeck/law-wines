```python
inventory = inventory.dropna(
    subset=["Order Number", "Sale Date"],
    how="all"
).copy()

inventory.shape
```

```python
inventory["Sale Date"] = pd.to_datetime(
    inventory["Sale Date"],
    errors="coerce"
)

inventory["Sale Date"].dtype
```

```python
print(inventory["Sale Date"].min())
print(inventory["Sale Date"].max())
print("Missing dates:", inventory["Sale Date"].isna().sum())
```

```python
inventory["Transaction Type"] = np.where(
    (inventory["Qty"] < 0) | (inventory["Net Revenue"] < 0),
    "Return/Adjustment",
    "Sale"
)

inventory["Transaction Type"].value_counts()
```

```python
negative_orders = inventory.loc[
    inventory["Order Number"] < 0,
    "Order Number"
].abs().unique()

inventory[
    inventory["Order Number"].isin(negative_orders)
][
    ["Order Number", "Sale Date", "Product Title", "Qty", "Net Revenue"]
].sort_values("Order Number")
```

```python
inventory[
    (inventory["Qty"] < 0) &
    (inventory["Net Revenue"] >= 0)
][
    ["Order Number", "Sale Date", "Product Title", "Qty", "Net Revenue"]
]
```

```python
inventory["Order Number"] = inventory["Order Number"].astype("Int64")
inventory["Customer No."] = inventory["Customer No."].astype("Int64")

inventory.dtypes
```

```python
inventory["Date"] = inventory["Sale Date"].dt.date
inventory["Month"] = inventory["Sale Date"].dt.to_period("M")
inventory["Year"] = inventory["Sale Date"].dt.year

inventory[
    ["Sale Date", "Date", "Month", "Year"]
].head()
```

```python
print("Order Source:")
print(inventory["Order Source"].value_counts(dropna=False))

print("\nShipping Method:")
print(inventory["Shipping Method"].value_counts(dropna=False))

print("\nWine Club Name:")
print(inventory["Wine Club Name"].value_counts(dropna=False))
```

```python
text_cols = [
    "Order Status",
    "Fulfillment Status",
    "Wine Club Name",
    "Cashier",
    "Order Source",
    "Shipping Method",
    "Bill State",
    "Product Title",
    "Part No"
]

for col in text_cols:
    inventory[col] = inventory[col].str.strip()
```

```python
def club_group(name):
    if pd.isna(name):
        return "Non-Club"

    if "White Label Waitlist" in name:
        return "White Label Waitlist"

    if "Black Label Waitlist" in name:
        return "Black Label Waitlist"

    if "White Label Allocation" in name:
        return "White Label Allocation"

    if "Black Label Allocation" in name or "Black Label Dinner" in name:
        return "Black Label Allocation/Special"

    if "White Label" in name:
        return "White Label"

    if (
        "Black Label" in name
        and (
            "3 btls Spring" in name
            or "6 btls Spring" in name
            or "12 btls Spring" in name
        )
    ):
        return "Black Label Spring"

    if "Black Label (12 bottles)" in name:
        return "Black Label 12 Only"

    if name == "Consumer":
        return "Consumer"

    return "Other"
```

```python
inventory["Club Group"] = inventory["Wine Club Name"].apply(club_group)

inventory["Club Group"].value_counts(dropna=False)
```

```python
inventory["Date"] = inventory["Sale Date"].dt.date
inventory["Month"] = inventory["Sale Date"].dt.to_period("M")
inventory["Year"] = inventory["Sale Date"].dt.year
```

```python
inventory["Club Group"].value_counts()
```

```python
inventory["Month Start"] = inventory["Sale Date"].dt.to_period("M").dt.to_timestamp()

inventory[
    ["Sale Date", "Month Start", "Qty", "Net Revenue"]
].head()
```

```python
inventory_monthly = (
    inventory
    .groupby("Month Start")
    .agg(
        Total_Quantity=("Qty", "sum"),
        Total_Revenue=("Net Revenue", "sum"),
        Number_of_Orders=("Order Number", "nunique"),
        Number_of_Customers=("Customer No.", "nunique"),
        Number_of_Transactions=("Order Number", "size"),
        Returns_Adjustments=("Transaction Type", lambda x: (x == "Return/Adjustment").sum())
    )
    .reset_index()
)

inventory_monthly
```

```python
inventory_monthly["Revenue_per_Order"] = (
    inventory_monthly["Total_Revenue"] /
    inventory_monthly["Number_of_Orders"]
)

inventory_monthly["Quantity_per_Order"] = (
    inventory_monthly["Total_Quantity"] /
    inventory_monthly["Number_of_Orders"]
)

inventory_monthly
```

```python
inventory_monthly_source = (
    inventory
    .groupby(["Month Start", "Order Source"])
    .agg(
        Total_Quantity=("Qty", "sum"),
        Total_Revenue=("Net Revenue", "sum"),
        Number_of_Orders=("Order Number", "nunique")
    )
    .reset_index()
)

inventory_monthly_source.head(20)
```

```python
inventory.to_csv("Inventory_Cleaned.csv", index=False)
inventory_monthly.to_csv("Inventory_Monthly.csv", index=False)
```

```python
inventory_monthly_source.to_csv(
    "Inventory_Monthly_by_Source.csv",
    index=False
)
```
